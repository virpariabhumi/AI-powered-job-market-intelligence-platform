**Dataset:** india_jobs_cleaned.csv (12,188 records)

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('../data/cleaned/india_jobs_cleaned.csv')
df.head()
```

## 1. Missing Values Overview

```python
missing = df.isnull().sum()
missing[missing > 0]
```

## 2. Top Job Designations

```python
plt.figure(figsize=(10, 4))
df['title'].value_counts().head(10).plot(kind='barh', color='teal')
plt.title('Top 10 Job Designations')
plt.xlabel('Number of Postings')
plt.gca().invert_yaxis()
plt.show()
```

## 3. Geographic Distribution (Top Locations)

```python
plt.figure(figsize=(10, 4))
df['location'].value_counts().head(10).plot(kind='bar', color='coral')
plt.title('Top 10 Job Locations')
plt.ylabel('Number of Postings')
plt.xticks(rotation=45, ha='right')
plt.show()
```

## 4. Key Skills Distribution

```python
skills = df['key_skills'].dropna().str.split(',').explode().str.strip().str.lower()
plt.figure(figsize=(10, 5))
skills.value_counts().head(15).plot(kind='barh', color='navy')
plt.title('Top 15 Skills in Market')
plt.xlabel('Frequency')
plt.gca().invert_yaxis()
plt.show()
```
